# 02 — Baselines and temporal evaluation

Never report only a random split if temporal data exists.

```python
from stresstest_fin.config import load_config
from stresstest_fin.data import load_and_validate, split_data, features_and_target
from stresstest_fin.models import build_models
from stresstest_fin.evaluate import predictive_metrics
cfg=load_config("../configs/example.yaml"); df=load_and_validate(cfg)
train,test,split_type=split_data(df,cfg)
Xtr,ytr=features_and_target(train,cfg); Xte,yte=features_and_target(test,cfg)
models=build_models(Xtr,cfg.random_seed)
reports={}
for name in cfg.models:
    model=models[name].fit(Xtr,ytr)
    reports[name]=predictive_metrics(yte, model.predict_proba(Xte)[:,1])
reports
```
